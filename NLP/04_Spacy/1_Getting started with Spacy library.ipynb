```python
import spacy
```

```python
import nltk
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
text = nlp('GFG is looking for Data science Inteerns')
```

### 1.Tokenization

```python
for token in text:
    print(token)
```

```python
text = nlp('The cost of Iphone in U.K is 699$')
```

```python
for token in text:
    print(token.text)
```

### 2.POS(Parts of Speech)

```python
for token in text:
    print(token.text, token.pos,)
```

```python
for token in text:
    print(token.text, token.pos_)
```

### 3.Sentance Tokenization

```python
txt = nlp('This is sentance one.this is second sentance.this is last one.tets study now')

for sent in txt.sents:
    print(sent)
```

```python
for sent in txt:
    print(sent)
```
