## 1.isalpha()

```python
text = 'Atul Pandey'
```

```python
def isalpha(text):
    c = 0
    for i in text:
        if((ord(i) >= 65 and ord(i) <= 90) or ord(i) >= 97 and ord(i) <=122):
            c += 1
    if(len(text)== c):
        return True
    else:
        return False
    
print('Atul Pandey'.isalpha())
print(isalpha('Atul Pandey'))
```

### 2.isdigit()

```python
def isdigit(text):
    c = 0
    for i in text:
        if((ord(i) >= 48 and ord(i) <= 57)):
            c += 1
    if(len(text)== c): 
        return True
    else:
        return False
    
print('1231'.isdigit())
print(isdigit('1234'))
```

### 3.isalnum()

```python
def isalnum(text):
    c = 0
    for i in text:
        if((ord(i) >= 65 and ord(i) <= 90) or (ord(i) >= 97 and ord(i) <=122) or (ord(i) >= 48 and ord(i) <= 57)):
            c += 1
    if(len(text)== c):
        return True
    else:
        return False
    
print('Atul Pandey'.isalnum())
print(isalnum('Atul Pandey'))
    
```
