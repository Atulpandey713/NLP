```python
text ="The quick brown fox jumps over the lazy dog"
```

### 1.string to ascii conversion

```python
ascii_=[{char:ord(char)} for char in text]
```

```python
ascii_
```

```python
asci_=[ord(chr) for chr in text]
```

```python
asci_
```

### 2.ASCII to String Conversion

```python
''.join([chr(asc) for asc in asci_])
```

```python
stri_ = ''.join([chr(ascii_[i][char]) for i in range(len(ascii_)) for char in ascii_[i]])
```

```python
stri_
```
