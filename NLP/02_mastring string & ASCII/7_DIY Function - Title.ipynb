```python
def upper(text):
    string = ''
    for char in text:
        if(ord(char) >= 97 and ord(char) <=122):
            string += chr(ord(char)-32)
        else:
            string += char
    return string

def lower(text):
    string = ''
    for char in text:
        if(ord(char) >= 65 and ord(char) <=90):
            string += chr(ord(char)+32)
        else:
            string += char
    return string
```

```python
def title(text):

    output = []
    for word in text.split():
        output.append(upper(word[0] + lower(word[1:])))
    return ' '.join(output)
```

```python
title('atul pandey')
```
