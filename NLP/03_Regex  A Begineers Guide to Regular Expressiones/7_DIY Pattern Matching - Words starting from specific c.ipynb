### 1.Word - Starting from specific Cgaracters

```python
txt = 'AI is having the capability to revolutionize the industries .The industry are ready for big change '
word = 'in'

for wrd in txt.split(' '):
    if (wrd[:2].lower() == word.lower()):
        print(wrd)
```

### 2.Word - Ending with specific Characters

```python
txt = 'AI is having the capability to revolutionize the industries . the industry are ready for big change '
word = 'he'

for wrd in txt.split(' '):
    if (wrd[-2:].lower() == word.lower()):
        print(wrd)
```
